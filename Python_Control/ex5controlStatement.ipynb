<h1 id="%EC%A0%9C%EC%96%B4%EB%AC%B8">제어문¶</h1>
while 기본 구조 이해 – 초기식, 조건식, 실행문, 증감식
무한 루프 활용 – while True:와 break로 반복 제어
리스트와 while문 활용 – 특정 값 존재 여부 확인 후 삭제
time 모듈 활용 – time.time()으로 현재 시간 확인, time.sleep()으로 실행 지연

```python
import time
import datetime
```

```python
comm="""
- 자바에서 배운 것 동일 함
1) 초기식
2) while 조건식:
3) 실행문
4) 증감식
"""
```

```python
# 초기식(i=0), 조건식(i<10), 실행문(print), 증감식(i+=1) 구조를 학습한다. 
# 기본 while 반복문
i = 0
while i < 10:
    print(f"{i} =>")
    #i++
    i +=1
print('*'*30)
```

```python
# while True: 로 무한 반복을 만들고, break로 종료 조건을 제어한다.
# 무한 루프 예제
while True:
    num = int(input('수입력:')) #blocking 함수
    if num % 2 == 0:
        print('짝수 입력')
    else:
        print('홀수 입력하면 끝남')
        break
print('*'*30)
```

```python
### 리스트와 while문 사용해보기 - while문 보다 for문이 가장 많이 사용
lista = list(range(1,10,2))
print(f"{lista}")
#리스트 안에서 3이란 숫자를 삭제 
e = 3
print(e in lista) #True
lista.remove(3)
print(e in lista) #False
print(f"{lista}")
```

```python
# <한번해보기>
# while 문으로 구현하기
# list가 있다면 조건문 => 3 in lista
# lista =[1, 3, 5, 7, 9, 3]
lista = list(range(1,10,2))
lista.append(3)
print(f"{lista}")
# 3이 존재하는지 확인
e = 3
print(e in lista)   # True

#3이란 값을 찾으면 계속 반복
while e in lista:
    #반복하면서 3만 삭제 
    lista.remove(e)
print(f"{lista}")
```

```python
# time.time()으로 현재 시간을 초 단위로 확인하고, time.sleep()으로 실행을 지연시킨다. 
# time 모듈
# time() => 현재 시간을 초 단위로 반환 (1970년 1월 1일 기준)
print(f"현재 시간(초) : {time.time()}")
# sleep() => 지정한 시간 동안 실행을 멈춤
print("5초 대기후 출력")
time.sleep(5)
print("대기 끝!")
```

```python
count = 5
while count > 0:
    print(count)
    time.sleep(1)
    count -= 1
print('종료')
```

```python
#짝수만 출력  1 ~ 10까지 
for i in range(1,11):
    if i % 2 != 0: # 홀수
        continue   # 아래의 실행문을 건너뛰고 다음 반복으로 이동해라!
    print(i)
```

```python
#while문으로 변경 해보기 
i = 1
while i < 10:
    i += 1
    if i % 2 != 0:   
        continue     
    print(i)
```

```python
# while이 정상 종료되면 else 블록 실행
i = 0
while i < 3:
    print(i)
    i += 1
else:
    print("while 정상 종료")
```

<h2 id="datetime-%EB%AA%A8%EB%93%88-***">datetime 모듈 ***¶</h2>
datetime.datetime.now()로 현재 날짜와 시간
.year, .month, .day 속성을 통해 날짜 값을 가져 오는 속성
.replace()를 사용해 날짜 객체를 수정
.date()를 사용해 시간 없이 날짜만 추출

```python
now = datetime.datetime.now()
print(f"현재 날짜와 시간: {now}")
print(f"년도 : {now.year}")
print(f"월 : {now.month}")
print(f"일 : {now.day}")
```

```python
# 2026년 8월 5일 일년 후 날짜 
now = datetime.datetime.now()
one_year_later = now.replace(year=now.year + 1)
print(f"일년 후 날짜:{one_year_later}")
```

```python
#datetime.timedelta를 활용해 날짜 간의 차이를 계산
daynum = 30
thirty_days_later = now + datetime.timedelta(days=daynum)
print(f"{daynum}일 후 날짜 :{thirty_days_later}")
```

```python
#날짜 차이
print(f"날짜로 부터의 경과일:{(now - datetime.datetime(2025,12,1)).days}")
```

```python
start = time.time()
count = 0
while time.time() - start < 3:
    count += 1
print(f"3초 동안 {count}번 반복했습니다.")
```

```python
#continue 문 
arr = list(range(0,30,3))
output = ""
for e in arr:
    if e % 2 == 0:
        continue
    output += str(e) + ","
print(output[:-1])
```

```python
dan = int(input('단을 입력하세요:'))
limits = int(input('범위를 입력하세요:[3,5,7,9]'))
check_limit = [3,5,7,9]
if check_limit.count(limits) <= 0:
    print(f"범위는 {check_limit} 중에 하나를 선택하세요")
else:
    start = int(dan - limits / 2) + 1
    end = int(dan + limits / 2) +1
    for x in range(0,10):
        for y in range(start,end):
            if x == 0:
                print(f"=={y}단==",end=" ")
            else:
                print(f"{y}x{x}={(x*y):2d}",end="  ")
        print()
```
