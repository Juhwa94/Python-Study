<h1 id="%EC%BB%B4%ED%94%84%EB%A6%AC%ED%97%A8%EC%85%98(Comprehension)">컴프리헨션(Comprehension)¶</h1>
리스트, 세트, 딕셔너리 등의 컬렉션을 빠르고 간결하게 생성할 수 있게 해주는 문법
여러 줄의 for 루프와 if 문을 한 줄로 줄여 가독성을 높임
일반적인 for 루프보다 실행 속도가 조금 더 빠름

```python
import sys
```

```python
#ex12_comprehension.py
# 리스트, 셋, 튜플, 딕셔너리 등을 반복문을 내제화해서 간단하게 만들 수 있는 문법
# 리스트 컴프리헨션의 예 => [ <실행문> for <변수> in <범위> if <조건> ]
```

```python
# 기본 for문으로 리스트 생성
lista = []
for i in range(0,20,5):
    lista.append(i * i)
print(lista)
```

```python
# 리스트 컴프리헨션으로 변경
lista = [i * i for i in range(0,20,5)]
print(lista)
```

```python
# 방금 위에서 설명한 대로 for문을 보고 대입 해보기 
array = []
for i in range(0, 20, 2):
    array.append(i + 2)
print(array)
# 리스트 컴프리헨션으로 변경
array1 = [ i + 2 for i in range(0, 20, 2)]
print(array1)
```

```python
# 하나 더 해보기: 알파벳 생성
# for문으로 ['a', 'b', 'c', 'd', 'e', 'f', 'g']
# ord('a') -> 97, chr(97) -> 'a'
# 리스트 컴프리헨션으로 변경
lista = []
for i in range(7):
    lista.append(chr(ord('a') + i))
print(lista)
```

```python
listb = ['바나나', '자두', '사과', '오렌지', '무', '체리']
```

```python
# 컴프리헨션 조건문 이해 
# -> [실행문 for 변수 in 반복가능객체 if 조건]
# -> 조건을 만족하는 경우만 리스트에 포함된다.
listb_res = [ e for e in listb if e != '무']
print(listb_res)
```

```python
# for문으로 '무' 제외한 후 listb_res = [] 에 저장해보기 
listb_res = []
for e in listb:
    #print(e)
    if e != '무':
        listb_res.append(e)
print(listb_res)
```

```python
# 한번 해보기 1>
# 1부터 20까지의 3의 배수를 list_arr에 저장한 후 출력하기
# 3의 배수 출력: [3, 6, 9, 12, 15, 18]
list_arr_comp = [x for x in range(1,21) if x % 3 == 0]
print(list_arr_comp)
```

```python
#한번 해보기 2> 2차원 좌표출력 : 2중 for문을 사용
# [ <실행문> for <변수> in <범위> if <조건> or for ]
list_matrix_comp = [(x,y) for x in range(1,4) for y in range(1,4)]
list_matrix_comp
```

```python
# 홀수 짝수 구분해서 출력 해보기 -> if else
# (조건이True실행) if 조건 else (조건이False실행)
# [(조건이True실행) if 조건 else (조건이False실행) for문 ]
list_even_odd = []
for x in range(1,6):
    if x % 2 == 0:
        list_even_odd.append(f"{x}짝수")
    else:
        list_even_odd.append(f"{x}홀수")
print(list_even_odd)
```

```python
list_even_odd2 = [f"{x}짝수" if x % 2 == 0 else f"{x}홀수" for x in range(1,6)]
print(list_even_odd2)
```

```python
#딕셔너리
# {'과일':단어수}
words = ['apple', 'banana', 'cherry']
word_len = {}
for w in words:
    print(f"{w}:{len(w)}")
    word_len[w] = len(w)
print(word_len)
#딕셔너리 컴프리핸션으로 변경 해보기 
word_len2 = { w:len(w)  for w in words}
print(word_len2)
```

```python
#Set - 
sets = set()
for c in "apple":
    if c != 'p':
        sets.add(c)
print(sets)
sets2 = {c for c in "apple" if c != 'p'}
print(sets2)
```

<h2 id="Generator">Generator¶</h2>
return 은 값을 반환 , yield는 값을 반환하고 잠시 정지 시킨다.

  <th>구분</th>

호출 방식
일반 함수 호출 => 결과값 바로 얻음
제너레이터 함수 호출 => 이터레이터 객체 반환


여러 값 반환
한 번만 반환 가능
여러 번 yield로 순차적으로 값 반환 가능


메모리 사용
모든 값을 한 번에 만들어 반환

```python
# return 문으로 반환 하는 경우 
def myfruit_res(fruits):
    result = []
    for f in fruits:
        result.append(f"맛있는 과일 =>{f}")
    return result
my_fruits = ['사과','바나나','딸기']
res1 = myfruit_res(my_fruits)
```

```python
# 반복문으로 출력 해보자.
for i in res1:
    print(i)
print('*'*30)
for i in res1:
    print(i)
```

```python
def myfruit_gen(fruits):
    result = []
    for f in fruits:
        result.append(f"맛있는 과일 =>{f}")
    yield result
my_fruits = ['사과','바나나','딸기']
res2 = myfruit_gen(my_fruits)
```

```python
# 반복문으로 출력 해보자.
for i in res2:
    print(i)
print('*'*30)
for i in res2:
    print(i)
```

```python
def myfruit_gen(fruits):
    for f in fruits:
        yield f"맛있는 과일  => {f}"
my_fruits = ['사과','바나나','딸기']
# yield , generator는 객체로 만들어서 주소값 반환
gen1 = myfruit_gen(my_fruits)
# 사용
list(gen1)
```

```python
# 재사용이 불가 
list(gen1)
```

```python
#generator는 () 사용해서 컴프리헨션으로 만드는 것이 일반적이다.
gen2 = (f"맛있는 과일  => {f}" for f in my_fruits)
```

```python
list(gen2)
```

```python
list(gen2)
```
