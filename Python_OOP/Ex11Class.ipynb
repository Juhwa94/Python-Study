```python
# setattr(self,'변수',값)
class Person2:
    # 생성자 정의 - 생성자 오버로딩을 지원하지 않는다. 
    # 생성자 매개변수  기본매개변수 초기화를 사용해서 해결 하기
    def __init__(self,nation=None,name=None,age=None,height=None,weigth=None):
        #인스변수에 등록 -> self.변수명=값
        # self.nation=nation
        # self.name=name
        # self.age=age
        # self.height=height
        # self.weigth=weigth
        setattr(self,'nation',nation)
        setattr(self,'name',name)
        setattr(self,'age',age)
        setattr(self,'height',height)
        setattr(self,'weigth',weigth)
        print("생성자가 초기화 되었음")
    #메서드 정의
    def display(self):
        print(f"국적:{self.nation}",end=",")
        print(f"나이:{self.age}",end=",")
        print(f"키:{self.height}",end=",")
        print(f"몸무게:{self.weigth}",end=",")
        print(f"이름:{self.name}",end="")
p1 = Person2('영국','프랑수아',20,177,68)
p1.display()
p2 = Person2('인도','테스형')
p2.display()
p3 = Person2()
p3.display()
```

```python
# 연습문제) 
# 클래스 Person2 의 소스를 참고해서 **kwargs로 생성자의 인자로 받고 
# 키와 값을 각각 인스턴스 변수로 등록 하시오
# 호출 값 : {"nation":"korea","name":"테스형","age":30,"height":177.7,"weigth":90.1}
# setattr(self,'변수',값)
class Person3:
    def __init__(self,**kwargs):
        for key, val in kwargs.items():
            print(f"{key} : {val}")
            setattr(self, key, val)
        #메서드 정의
    def display(self):
        print(f"국적:{self.nation}",end=",")
        print(f"나이:{self.age}",end=",")
        print(f"키:{self.height}",end=",")
        print(f"몸무게:{self.weigth}",end=",")
        print(f"이름:{self.name}",end="")
p1 = Person3(**{"nation":"korea","name":"테스형","age":30,"height":177.7,"weigth":90.1})
p1.display()
```

<h2 id="%EC%A0%95%EC%A0%81%EB%A9%94%EC%84%9C%EB%93%9C-(-static-)">정적메서드 ( static )¶</h2>
생성없이 클래스영역과 무관하게 독립적으로 사용하는 영역
@staticmethod 사용했을 때 와 그렇지 않을 때의 차이를 안다.

```python
class MyStaticClass:
    num = 10
    #static메서드로 적용
    @staticmethod
    def add(x,y):
        #static영역에서는 인스턴스를 생성없이 참조 할 수 없다.
        #NameError: name 'self' is not defined
        #self.age = x + 10
        print(f"클래스 변수 : {MyStaticClass.num}")
        return x + y
    def printX(self):
        print(self.num,":",MyStaticClass.num)
ref = MyStaticClass()
print(ref.add(10,10))
ref.printX()
```

```python
# 연습문제) 주어진 UML을 보고 똑 같이 구현 해봅시다.
class MyStaticNum:
    #private static int num2;//python 클래스변수
    num2 = 0
    #private int num1; // 인스턴스 변수 self.num1
    def __init__(self):
        self.num1 = 0
    """
    //python @staticmethod
	public static void staticAdd() {
		num2 += 1;
	}
    """
    @staticmethod
    def staticAdd():
        MyStaticNum.num2  = MyStaticNum.num2 + 1
    """
	public  void nonstaticAdd() {
		num1  = num1 + 1;
	}
    """
    def nonstaticAdd(self):
        self.num1  = self.num1 + 1
    def printStatic(self):
        return f"static_num2:{MyStaticNum.num2} , nonstatic_num1:{self.num1}"
```

```python
t1 = MyStaticNum()
t2 = MyStaticNum()
```

```python
t1.staticAdd()
t1.nonstaticAdd()
t2.staticAdd()
t2.nonstaticAdd()
print('*'*30)
print(t1.printStatic())
print(t2.printStatic())
```

<h2 id="@classmethod">@classmethod¶</h2>
<p>클래스 메서드는 클래스 안에서 클래스 변수를 해당 함수에서 사용하려고 할 때 선언</p>

<p>cls 인자로 클래스 변수를 참조하기 위한 주소 값을 받아서 사용한다</p>

```python
class Ractangle:
    count = 0  #class변수
    def __init__(self,w,h):
        self.w = w
        self.h = h
        self.count = 10 # 인스턴스 변수
        print('생성자 초기화')
    def calArea(self):
        result = self.w + self.h
        return result
    
    @staticmethod    
    def isSeq(rW,rH):
        return rW == rH
        
    @classmethod
    def printRect(cls):
        print(f"CLS => {cls.count} , class변수 : {Ractangle.count}")
    
    def selfTest(self):
        print("selfTest 에서 class변수 호출")
        print(f" Class변수값 : {Ractangle.count}")
        return self.count #인스턴스 값 10을 반환
```

<h2 id="%EC%83%81%EC%86%8D-*****">상속 *****¶</h2>
<p>부모의 자원을 확장, 재사용</p>

<p>class Class(SuperClass):</p>

<p>Python은 다중 상속이 가능하다.class Class(SuperClass1,SuperClass2):</p>

<p>super() : 부모 클래스의 주소</p>

```python
class Calculator:
    def __init__(self,first=10,second=0):
        print("부모 생성자 호출")
        self.first=first
        self.second=second
    def setData(self,first,second):
        self.first=first
        self.second=second
    def sum(self):
        return self.first + self.second
    def mul(self):
        return self.first * self.second
    def sun(self):
        return self.first - self.second
    def div(self):
        return self.first / self.second
```

```python
ref = Calculator(10,20)
ref.sum()
```

```python
# 부모 클래스를 상속 받는 자식 클래스 정의
# class ClassName(부모클래스1, 부모클래스2)
class Calc2(Calculator):
    def __init__(self,cc=10):
        # 상속 받은 부모 클래스의 생성자를 호출한다.
        super().__init__(2,5)
        print("자식 생성자를 호출")
        #자식의 인스턴스 변수에 값을 전달 
        self.cc = cc
    def pow(self):
        #상속관계에서 self는 부모의 자원을 가리킨다. 이때 자식이 중복된 것이 없으면 .....
        result = self.first ** self.second
        return result
```

```python
#자식 클래스 생성
b = Calc2()
print(b.sum()) #부모
print(b.mul()) # 부모
print(b.pow()) # 자식
print(b.cc) #자식
```
