```python
class Person:
    #클래스 속성 
    category = "Human"
    #생성자
    def __init__(self,name,age):
        self.name = name
        self.age = age
print(f"클래스 속성 : {Person.category}")
# print(f'인스턴스 속성 : {Person.name}')
#클래스를 생성후 사용
p1 = Person('테스형',200)
p2 = Person('코스형',100)
# 인스턴스는 주소마다 각각
print(f'p1 주소 => {p1.name}') 
print(f'p2 주소 => {p2.name}')
print(p1.category,p2.category) # 클래스 속성은 주소가 달라도 공유!
print(f"주소값 비교해보기 {p1 == p2}")
print(f" p1 , p2 의 주소 :  {id(p1)} , {id(p2)} ")
```

```python
# 클래스를 만드는 일련에 과정 
# 클래스를 정의하기 전에 일반적인 개발을 하다가 ~~
# 아!!!!! 이것은 재사용할 수 있겠다. -> 함수를 정의 해볼 수 있다.
# num1 = 0
# num2 = 10
# resv = num1 + num2
# print(resv)
```

```python
# 함수화 => 클래스의 메서드로 변경 
# 함수화 => 클래스의 메서드로 변경 
def test():
    return "ict01기 Hi~"
def oper1(num1,num2):
    resv = num1 + num2
    msg = test()
    return resv  
def oper2(num1,num2):
    resv = num1 - num2
    msg = test()
    return resv 
def oper3(num1,num2):
    resv = num1 * num2
    return resv
# 함수 테스트 
print(oper1(10,10))
print(oper2(11,10))
print(oper3(2,10))
```

```python
# 함수와 메서드의 차이
# 메서드 : (self) , class 내에 정의된 함수를 의미 
# 클래스로 정의하기
# ctrl + ] : =>,  ctrl + [ : <= (들여쓰기 단축키)
class OperClass:
    def test(self):
        return "ictAI 01기 Hi~^^"
    def oper1(self,num1,num2):
        resv = num1 + num2
        msg = self.test()  #인스턴스 메서드가 호출
        print("메서드 안에서 호출:",msg)
        return resv  
    def oper2(self,num1,num2):
        resv = num1 - num2
        msg = self.test()
        print("메서드 안에서 호출:",msg)
        return resv 
    def oper3(self,num1,num2):
        resv = num1 * num2
        return resv
```

```python
s1 = OperClass()
s1 = OperClass()
print(type(s1))
print(s1.oper1(10,20))
print(s1.test())
print("*"*30)
print(s1.oper2(20,10))
```
