```python
t1 = (1,2,'a','b','c')
print(type(t1))
print(t1)
```

```python
# 리스트와 동일하게 인덱스로 접근 가능하지만 값 변경은 불가능하다.
print('index를 사용해서 값을 가져 오기')
print(t1[0]) #1
print(t1[-1]) #c
```

```python
#TypeError: 'tuple' object does not support item assignment
# 튜플이기 때문에 item에 값이 변경이 안됨!
#t1[0] = 'd'
#tuple이기 때문에 삭제도 안된다.
#del t1[2]
```

```python
# 튜플 연산 
# 튜플은 새로운 튜플을 생성하는 방식으로 연산된다.
# tuple의 연산 - 새로운 tuple로 생성이 된다. 
t1 = (1,2,'a','b','c')
t2 = (5,6)
print(f't1_1 => {id(t1)}') # t1의 주소값을 출력 
t1 = t1 + t2
print(f't1_2 => {id(t1)}') # 연산이 된 t1의 주소값을 출력 
print(id(t1*2)) # 연산시에 새로운 객체로 생성이 된다를 증명 !!!
print(f't1_3 => {id(t1)}')
```

```python
#슬라이싱 
print('tuple의 슬라이싱')
t1 = (1,2,'a','b','c')
print("t1:",t1,"id=>",id(t1))
print(t1[1:], id(t1[1:])) # index 가 1 부터 슬라이싱 해서 새로운 tuple 로 만듬
print(t1[1:3])
```

```python
# tuple에서 원소를 뽑아서 변수로 각각 저장  *****
t1 = (1,2,3)
n1,n2,n3 = t1
print(f"{n1},{n2},{n3}")
#format함수로 출력
print("{},{},{}".format(*t1))
```

```python
# range()함수를 사용해서 튜플을 생성시 동적으로 만들기
t1 = tuple(range(1,10))
print("{},{},{},{},{}".format(*t1))
```

```python
# join함수
# 튜플 요소를 문자열로 변환 후 연결
# join함수 : list,tuple에서 문자열로 요소를 변경해서 
lista = ['1','2']
stri = '*'.join(lista)
print(stri)
```

```python
t1 = tuple(range(1,10))
print(type(t1[0]))
#str함수로 변환 
#join(x실행 for x in tuple/list)
res = ','.join(str(x) for x in t1)
res
```

```python
reslist = []
for x in t1:
    reslist.append(str(x))
res = ','.join(reslist)
res
```

```python
#res를 # 리스트로 변경하시오.~ 
lista = res.split(',')
print(lista)
```

```python
#swap 알고리즘 -  변수의 값을 교환 
t1 = (1,2,3)
n1,n2,n3 = t1
## 변수의 값을 교환 
n1,n2 = n2,n1
print("{},{}".format(n1,n2))
```

```python
# tuple의 값을 변경 해야 할 경우?  tuple -> list -> tuple 이런식으로 해야 함!
my_list = [1,2,3]
my_tuple = ('a','b','c')
#list => tuple
a = tuple(my_list)
print(a)
b = list(my_tuple)
print(b)
```

```python
# 제네레이터 방식을 사용하면 주소를 보내기 때문에 속도가 개선이 되는데
# 이때 제네레이터 방식을 튜플로 만든다. (데이터 보호 lock 개념 )
print(type(a),type(b)) 
my_tuple = tuple(b)
print(my_tuple)
```

```python
# 알파벳 함수를 사용하지 않고 컬랙션 자료형을 생성한다면
# lista = []
# seta =  {} 
# mytuple = () 
# 연습문제 위의 방식으로 각요소의 값이 1만 저장하도록 생성한 후 출력 해보기 
lista = [1]
seta = {1}
#***** # 요소가 하나일 때는 반드시 콤마 필요
mytuple = (1,) 
print(type(lista),lista)
print(type(seta),seta)
print(type(mytuple),mytuple)
```

```python
# 튜플은 불변이지만 count(), index() 메서드는 제공
t2 = (1,2,3,2,2) 
print("count(2):", t2.count(2))
print("index(3):", t2.index(3))
```

```python
# 튜플 안에 튜플을 넣고 접근하는 방법
t3 = (1,(2,3),(4,5))
print("중첩 튜플 접근:", t3[1][0]) # 2
```

```python
#딕셔너리 {"key":val}
#TypeError: unhashable type: 'list'
# 튜플은 불변이라 딕셔너리 키로 사용 가능
# d = {["Alice","Xman"]:"친구"}
d = {("Alice","Xman"):"친구"}
print("딕셔너리 키로 튜플 사용:",d[("Alice","Xman")])
```

```python
# 여러 값을 하나의 튜플로 묶거나 풀어내는 문법
packed = 1,2,3 # 튜플 
print(f"패킹된 튜플:{packed} , {type(packed)}")
a,b,c = packed
print(f"언패킹 {a},{b},{c}")
```
