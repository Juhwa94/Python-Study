```python
#Boolean(불리언) 자료형은 True와 False 두 값만 가진다.
#조건문(if/else)에서 불리언 값은 흐름 제어에 사용된다.
#0 : False
#0이 아닌 모든 수 : True
#비교 연산자(==, !=, <, >, <=, >=)의 결과는 항상 불리언 값
#논리 연산자 (not : 반대 , and : 둘 다참 , or : 둘 중하나라도 참 )
```

```python
"""
 Boolean
 불린 / 불리언 / 불
 True와 False 값만 가질 수 있음
"""
print(True)
print(False)
```

```python
# if else 
if True:
    print("True 입니다")
else:
    print("False 입니다")
```

```python
if 1:
    print("상수 1은 파이썬에서는 True 입니다.")
else:
    print("상수 1은 파이썬에서는 False 입니다.")
```

```python
if 0:
    print("상수 0은 파이썬에서는 True 입니다.")
else:
    print("상수 0은 파이썬에서는 False 입니다.")
```

```python
print(10 == 100)
print(10 != 100)
print(10 < 100)
print(10 > 100)
print(10 <= 100)
print(10 >= 100)
```

```python
#문자열에서도 비교연산이 적용이 가능한데
#한글일 때는 사전순(가나다라순) 앞에있는 갚이 작은 값이다.
print("가방" == "가방")
print("가방" != "하마")
print("가방" < "하마")
print(f'가방 > 하마 즉, 한글은 앞에 단어가 작은 값이다. => {"가방" > "하마"}') # False
```

```python
# not 연산자
print(f'not True => {not True}')
print(f'not False => {not False}')
```

```python
#아래의 비교연산은 불가능
# print(!True)
```

```python
# not연산자 조합
x = 10
under_20 = x < 20
print(f"under_20 : {under_20}") # True
print(f"not under_20 : {not under_20}")  # False
```

```python
# and , or
print(True and True)   # True
print(True and False)  #False
print(False and True)  #False
print(False and False) #False
print(True or  True)  # True
print(True or  False) # True
print(False or  True) # True
print(False or  False) #False
```

```python
#ex8_if.py
#1 - 기본 if문 
#사용자가 입력한 숫자가 양수인지 확인하는 프로그램을 작성하세요.
#입력 5 -> 출력 : 양수입니다.
#입력 -3 -> 출력 없음 
num = int(input("숫자를 입력하세요: ")) 
if num > 0: 
    print("양수입니다.")
```

```python
### 2. if-else 문
# 사용자가 입력한 숫자가 짝수인지 홀수인지 판별하는 프로그램을 작성하세요. % 연산자 사용 
num = int(input("숫자를 입력 "))
if num % 2 == 0:
    print(f"{num} 짝수")
else:
    print(f"{num} 홀수")
```

```python
# 3. if-elif-else 문
# 시험 점수를 입력받아 학점(A~F)을 출력하는 프로그램을 작성하세요.
# 90점 이상: "A 학점"  =>  score >= 90 
# 80점 이상: "B 학점"
# 70점 이상: "C 학점"
# 60점 이상: "D 학점"
# 그 외: "F 학점"
score = int(input("점수를 입력하세요: "))
if score >= 90:
    print("A 학점")
elif score >= 80:
    print("B 학점")
elif score >= 70:
    print("C 학점")
elif score >= 60:
    print("D 학점")
else:
    print("F 학점")
```

```python
#사용자가 입력한 나이에 따라 영화 관람료를 출력하는 프로그램을 작성하세요.
"""
0~12세: "어린이 요금 5000원"
13~18세: "청소년 요금 7000원"
19세 이상: "성인 요금 10000원"
"""
age = int(input("나이를 입력하세요: "))
if age <= 12:
    print("어린이 요금 5000원")
elif age <= 18:
    print("청소년 요금 7000원")
else:
    print("성인 요금 10000원")
```

```python
#사용자가 입력한 숫자가 양수/음수/0인지 판별하는 프로그램을 작성하세요.
"""
입력: 10 => 출력: "양수입니다."
입력: -5 => 출력: "음수입니다."
입력: 0 => 출력: "0입니다."
"""
num = int(input("숫자를 입력하세요: "))
if num > 0:
    print("양수입니다.")
elif num < 0:
    print("음수입니다.")
else:
    print("0입니다.")
```
