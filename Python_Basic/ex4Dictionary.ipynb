```python
#딕셔너리 기본 생성 및 출력
dict_a = {
    "name":"오징어 게임3",
    "type":"생존 스릴러",
    "price":30000
}
print(type(dict_a),dict_a)
```

```python
#딕셔너리 요소 접근 : 딕셔너리 출력  => 딕셔너리변수['키값']
print("영화 제목:",dict_a['name'])
print("장르:",dict_a['type'])
print("가격:",dict_a['price'])
```

```python
msg="""
f,format 함수를 사용해서 출력 해보기 
"""
print(msg)
print(f"영화제목 : {dict_a['name']} , 종류 : {dict_a['type']},price : {dict_a['price']}")
print("영화제목 : {} , 종류 : {} , price : {}"
      .format(dict_a['name'],dict_a['type'],dict_a['price']))
```

```python
print("*dict.values()함수 적용해보기")
print(type(dict_a.values()))
print(dict_a.values())
print(*dict_a.values())
print("영화제목 : {} , 종류 : {}, price:{}".format(*dict_a.values()))
```

```python
# 딕셔너리는 다양한 자료값을 가질 수 있으며, 특히 자료의
# 집합인 리스트를 가질 수 있다.
dict_b = {
    'director':["안소니 루소","조 루소"],
    'cast':["아이언맨","토르","타노스","닥터스트레인지","헐크"]
}
print(dict_b)
#안소니 루소를 출력해보시오.(한번 해보기)
print(dict_b["director"])
print(dict_b["director"][0])
```

```python
print("#특정 위치의 인덱스로 값에 접근 하는 방법 ")
idx1 = dict_b['cast'].index('토르') # 1
idx2 = dict_b['cast'].index('닥터스트레인지') #3
print(idx1,idx2)
```

```python
#<한번 해보기>
#"토르","타노스","닥터스트레인지"를 출력 해보시오.
# 이미 구한 idx1, idx2를 사용할 것 - 슬라이싱!
type(dict_b['cast'])
print(dict_b['cast'][idx1:idx2+1])
```

```python
print('딕셔너리 요소에 접근 해보기 실습')
dict_c = {
    "name":"ICT03 Study Python",
    "type":"book",
    "team":["ATeam","BTeam","CTeam"],
    "office":"신촌"
}
print("name :",dict_c["name"])
# value 
# 딕셔너리에 값을 추가하거나 변경할 때 
# 딕셔너리['기존키'] = '변경될 값' 
# 딕셔너리['새로운키'] = '새로운 값 '
dict_c["name"] = 'ICT Study AI 활용'
dict_c['pay'] = 100000
print(dict_c)
```

```python
## 딕셔너리 함수 
# get(key), keys(),values()
print('딕셔너리 함수 ')
dict_e = dict()
print(type(dict_e),":",dict_e)
dict_e = {}
print(type(dict_e),":",dict_e)
# {1} 한개의 원소라도 존재하면 set
# {} => 비워져 있으면 기본 딕셔너리로 생성이 된다.
# {1} => 한개의 원소라도 존재하면 set 으로 생성이 된다.
sets = {1}
sets.add(10) # set에서 제공하는 메서드
print(type(sets),":",sets)
print('*'*30)
```

```python
# key는 frozenset구조 -> 이거쓴 자바의 Map구조와 동일함.
# 새로운 키를 추가하면 기존의 딕셔너리에 추가되는 개념
dict_e = {}
dict_e['name'] ='테스형'
dict_e['head'] = '테스형 머리'
dict_e['body'] = '테스형 몸통'
print(f'요소 추가 이후 :{dict_e}')
dict_e['name'] ='테스형2'
dict_e['head'] = '테스형 머리2'
dict_e['body'] = '테스형 몸통2'
print(f'요소 추가 이후 :{dict_e}')
print('get(key) : 요소 가져오기 ')
print(f"get(key) : {dict_e.get('name')}")
print(f"dict_e['name'] => {dict_e['name']}")
```

```python
# keys(),values() 는 dict_keys,dict_values 클래스안에 <리스트>로 담아서 반환한다. ***** 
print(f"keys => {dict_e.keys()}") # dict_keys(['name', 'head', 'body'])
# type 확인 해보기
print(f"keys type => {type(dict_e.keys())}")
print("*"*30)
# 같은 형식으로 values도 확인 해보기 
print(f"keys => {dict_e.values()}")
print(f"values type => {type(dict_e.values())}")
```

```python
dict_e.keys()
dict_key_list = list(dict_e.keys())
print(dict_key_list)
```

```python
for e in dict_e.keys():
    print(e,end=' ')
print('*'*30)
for e in dict_e.values():
    print(e,end=' ')
print('*'*30)
```

```python
#items() 활용 
#이제 제일 중요한거 ~~~~~~~~~~~~~
# items() => key,value 를 list안에 tuple에 담아서 반환  ******
# dict_items([('name', '테스형2'), ('head', '테스형 머리2'), ('body', '테스형 몸통2')])
print(dict_e.items())
```

```python
print(list(dict_e.items())[0] ,":",type(list(dict_e.items())[0]))
print(list(dict_e.items())[1] ,":",type(list(dict_e.items())[1]))
```

```python
# 키와 값을 추출하려면?  아까 튜플에서 배운거 대로 사용함 (위와 동일함 )
b = list(dict_e.items())
print(b[0])
key_a, val_b = b[0]
print(f'Key : {key_a} , Val : {val_b}')
print('*'*30)
#tuple로 (key,value) 반환하니까 변수도 2개를 나열해서 받을 수 있다.
for i,v in dict_e.items():
    print(i,":",v)
```

```python
dict_f = {
    "brand":"Ford",
    "model":"Mustang",
    "year": 1996
}
print(dict_f)
```

```python
#pop(key), popitem()
#pop(key) : 키값을 가지고 지정한 항목을 꺼낸 후 삭제
#popitem() : 마지막 요소를  꺼낸 후 삭제
print(dict_f.pop("model"))
print(dict_f)
print(dict_f.popitem())
print(dict_f)
```
