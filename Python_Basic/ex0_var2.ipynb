```python
#i*10**-i :거듭제곱 연산 
#float(f"1e-{i}") : 문자열 "1e-3" 같은 지수 표기 문자열을 float()로 변환하는데 , 
#작은 수(10^-1 ~ 10^-4)는 일반 소수 표기법으로 출력되는데
#더 작은 수(10^-5 이하)는  1e-05, 1e-06, … 처럼 지수 표기법으로 자동 변환된다 *****
for i in range(10):
    print(i,"=>",1*10**-i,end="=======>")
    print(i,"=>",float(f"1e-{i}"))
#5 => 1e-05=======>5 => 1e-05  ============> e-05 는 10^-5 = 0.00001이다.
```

```python
#ex2_datatype2.py
########### 8진수 
##<참고>
# 머신러닝/딥러닝에서는 별로 안쓴다.
# 리눅스에서  chmod 755 같은 명령은 8진수로 권한을 표현 한 것이다.
# 권한을 숫자로 표현할 때, 각 그룹의 권한을 3비트(binary)
# 읽기 = 100 (2진수) = 4 (8진수) ->r
# 쓰기 = 010 (2진수) = 2 (8진수) ->w
# 실행 = 001 (2진수) = 1 (8진수) ->x
######################################################
# -> 8진수는 0부터 7까지 숫자만 사용하는 수
# 파이썬에서는 0o 또는 0O(숫자 0 + 알파벳 o/O)로 시작하면 8진수임
print("8진수(ocatal) : 숫자 0 + 알파벳 소문자/대문자 o")
print(f"0o177 => {0o177}") # 127
# 1 x 8^2 + 7 x 8^1 + 7 = 127
print("파이썬에서 10진수로 연산 해서 출력 해보기 ")
print(1 * 8**2 + 7 * 8 + 7) # 127
print(f"127을 8진수로 변경하는 함수 oct(127) =>{oct(127)}")
```

```python
print("""
      16진수(hexadecimal) : 숫자 0 + 알파벳 대소문자 x [0  ~ 9, A, B, C, D, E, F]
      """)
print(f"0xA => {0xA}") # 10
print(f"0x2F => {0x2F}") # 47
"""
0x2F = (2 x 16^1) + (F x 16) 
     = 32 + 15 = 47 
"""
# 16진수 → 10진수
print(f"int(2F,16) => {int('2F', 16)}") # => 47
# hex(10진수) =>  10진수 → 16진수 
print(f" hex(10진수) =>  {hex(47)}")  #0x2f
```

```python
#정수 출력의 다양한 형태 - 개념만 
##정수를 특정 칸에 출력하기
output_a = "{:d}".format(52)
print("정수값 출력 : d => ",output_a)
#특정 칸에 출력 - 특정 칸에 맞춰서 숫자를 출력하는 형태
output_b = "{:5d}".format(52)
print("정수값 출력 : d => ",output_b)
output_c = "{:10d}".format(52)
print("정수값 출력:d =>",output_c)
print("*"*30)
# 0000으로 채우기 - 빈칸을 0으로 채우는 형태
output_d = "{:05d}".format(52) #양수
print("정수값 출력:d =>",output_d) # 00052
output_e = "{:05d}".format(-52) #음수
print("정수값 출력:d =>",output_e) # -0052
print("*"*30)
```

```python
#기호 붙여 출력하기
# {:+d}: d 앞에 + 기호 추가하면 양수의 경우 + 붙여줌
# {: d}:  d 앞에 공백을 두면 양수의 경우 기호 위치를 공백으로 비워줌
output_f = "{:+d}".format(52) #양수
output_g = "{:+d}".format(-52) #음수
output_h = "{: d}".format(52)
output_i = "{: d}".format(-52) # 음수: 기호 부분 공백
print("# 기호와 함께 출력하기")
print(output_f)
print(output_g)
print(output_h)
print(output_i)
```

```python
# 다양한 형태의 부동 소수점 출력하기
print(f"부동 소수점 출력")
output_a = "{:f}". format(52.273)
output_b = "{:15f}". format(52.273) # 15칸 만들기
output_c = "{:+15f}". format(52.273) # 15칸에 부호 추가하기
output_d = "{:+015f}".format(52.273) # 15칸에 부호 추가하고 0으로 채우기
print(output_a)
print(output_b)
print(output_c)
print(output_d)
#소수점 아래 자릿수 지정하기 *****
print('*'*30)
output_a = "{:15.3f}". format(52.273)
output_b="{:15.2f}". format(52.273)
output_c="{:15.1f}". format(52.273)
output_d="{:15.0f}". format(52.273)
print(output_a)
print(output_b)
print(output_c)
print(output_d)
print(type(float(output_d)),":",float(output_b))
```

```python
#문자열 찾기: find()와 rfind()
#find() - 왼쪽부터 찾아서 처음 등장하는 위치 찾음 - index를 반환한다.
#rfind() - 오른쪽부터 찾아서 처음 등장하는 위치 찾음
msg2 = "안녕안녕하세요".find("안녕")
msg3 = "안녕안녕하세요".rfind("안녕")
print("안녕 find => {}".format(msg2)) # index 0
print("안녕 find => {}".format(msg3)) # index 2
```

```python
#f-문자열 - 많은 데이터일 경우는 속도가 빠르고, format함수에 비해서는 가독성 별로다.
# -파이썬 3.6버전부터 사용 가능
# - 문자열 앞에 f를 붙이고 문자열 내부에 표현식을 {} 괄호로 삽입
print("{}".format(10*2))
print(f"{(10*2)}")
#변수로 분리 되었을 때 상당히 편하다
num1 = 10*2
print("{}".format(num1))
print(f"{(num1)}")
print(f" 3 + 10 = {3 + 10}")
#""" 문자열로 저장시 변수와 함께 사용하기도 편하다.
msg2 = f"""
1+2 = {1+2}
2+3 = {2+3}
"""
print(msg2)
print("*"*30)
```

```python
#잠깐 리스트 살펴보기
#리스트는 자바의 배열과 달리 - 같은 자료형이 아니어도 생성 됨!
data = ['별',2,'M','서울시 강남구','Yes']
print(type(data))#list 
##list도 시퀀스 인덱스 0번 부터 
print(f"{type(data[0])}:{data[0]}")
print(f"{type(data[1])}:{data[1]}")
#전개연산자 - 리스트를 전개
print(data)
#문자열로 전개, 매개변수 인자안에서 사용, aa(String ... vars)
print(*data)
print(data[0],data[1])  #손이 아파서~~~
```
