```python
#문제3) 다음 문장의 단어에 빈도수를 구해서 딕셔너리로 아래처럼 출력하시오.
# 이때 반복되는 단어의 수가 1자리면 생략하고 , 
# '여러분', '매우','유용','해야', '에서', '합니다','입니다' 란 단어에 해당되면 제거 한다.

# """
# sentence = """
# 파이썬 ict패스워드 여러분 
# 파이썬 매우 강력한 프로그래밍 언어 입니다
# 파이썬 데이터 분석 매우 유용 합니다
# ict패스워드 여러분 파이썬 공부 매우 열심히 해야 합니다
# 파이썬 공부 ict패스워드 에서 핵심 과목 입니다
# 데이터 분석 반드시 파이썬 공부 중요 합니다
# """
sentence = """
파이썬 ict패스워드 여러분 
파이썬 매우 강력한 프로그래밍 언어 입니다
파이썬 데이터 분석 매우 유용 합니다
ict패스워드 여러분 파이썬 공부 매우 열심히 해야 합니다
파이썬 공부 ict패스워드 에서 핵심 과목 입니다
데이터 분석 반드시 파이썬 공부 중요 합니다
"""
# 조사, 의미 없는 단어 등을 미리 정의
stopwords = {'여러분', '매우','유용','해야', '에서', '합니다','입니다'}
## 문장을 공백 기준으로 잘라서 단어 리스트로 만듦
words = sentence.split()
# 단어별 등장 횟수를 저장할 딕셔너리
# 형식: {단어: 빈도수}
word_freq = {}
# 모든 단어를 하나씩 확인해서 word_freq에 넣는다.
for word in words:
    # 불용어에 포함되지 않은 단어만 처리
    if word not in stopwords:
         # 이미 있는 단어라면 , 기존 빈도수에 1 증가
         if word in word_freq:
            word_freq[word] += 1
         else:
            # 첫 단어라면 빈도수 1로 등록 
            word_freq[word] = 1 
# 2번 이상 등장한 단어만 담아둘  딕셔너리             
filter_freq = {}
for word, freq in word_freq.items():
    # 등장 횟수가 1보다 큰 단어만 선택
    if freq > 1:
        filter_freq[word] = freq
print(filter_freq)
```

```python
dan = int(input("단을 입력하세요: "))
num = int(input("범위를 입력하세요:[3,5,7,9]: "))
start = int(dan - (num - 1) / 2)
end = int(dan + (num - 1) / 2)

# 각 한 칸(열)의 최대 폭을 정의
# 예: "101x9=909" 문자열의 전체 길이를 계산
long_expr = f"{end}x9={end * 9}"
cell_width = len(long_expr)

for i in range(0, 10):
    for j in range(start, end + 1):
        if i == 0:
            # 단 제목도 데이터 열 폭에 맞춰 정렬시키는데  {:^10} => Center-align in width 10
            print(f"{'==' + str(j) + '단==':^{cell_width}}", end="  ")  
        else:
            # 식 전체를 왼쪽 정렬(<)하거나 오른쪽 정렬(>)하여 폭을 맞출수 있다.
            #{:>10} => Right-align in width 10
            print(f"{f'{j}x{i}={j*i}':<{cell_width}}", end="  ")  
    print("")
```
