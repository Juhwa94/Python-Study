<p>requests 모듈이란?</p>
<p>requests 모듈은 파이썬에서 HTTP 요청(GET, POST 등) 을 쉽게 보낼 수 있도록 도와주는 라이브러리이다.
주로 웹 서버에 요청을 보내고, 응답으로 전달되는 HTML 문서를 수집할 때 사용한다.</p>




데이터 수집 흐름
requests 모듈을 사용해 서버에 요청
서버로부터 HTML 응답을 받음
응답 데이터(response.text)에서 필요한 정보 추출
문자열 처리 또는 파싱을 통해 데이터 활용



수집 대상 사이트
URL: https://www.hanbit.co.kr/media/
수집 대상: 게시판 정보 및 메타데이터




GET 방식 요청과 파라미터 전달
GET 방식은 URL 뒤에 파라미터를 붙여서 서버에 데이터를 전달한다.

<p>예시</p>
import requests

params = {
    "page": 1
}

response = requests.get(

```python
# 실습하는 한빛 사이트와 같이 워낙 스크레핑을 많이 당해서
# 봇 (CDN으로 봇/스크래핑을 막아서 403을 에러가 난것이다.
print(f"{type(res.content)}")
print(res.content)
print(res.text[0:500])
```

```python
#브라우저 콘솔에서 코딩
#navigator.userAgent 
#Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36 OPR/133.0.0.0
```

```python
url = 'https://www.hanbit.co.kr/media/'
headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36 OPR/133.0.0.0",
    "Accept-Language":"ko-KR,ko;q=0.9",
    "REferer":"https://www.hanbit.co.kr/"
}
res = requests.get(url,headers=headers,timeout=10)
html = res.text[:800]
print(html)
```

```python
pos = html.find('og:description')
print(f"index:{pos}")
```

```python
if pos != -1:
    chunk = html[pos:pos+400]
    #print(chunk)
    content = chunk.split('content=')[1].split('"')[1]
    print(content)
```
