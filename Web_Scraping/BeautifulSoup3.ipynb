```python
# 지금까지 한 내용을 기준으로 함수로 정의 
#lognum = 0
def loopkyoboBooks(searchVal,page=1):
    print(f"Page Num => {page}")
    url = f"https://search.kyobobook.co.kr/search?keyword={searchVal}&target=total&gbCode=TOT&page={page}"
    res = requests.get(url)
    soup = BeautifulSoup(res.text,'html.parser')
    s = soup.select('div.auto_overflow_contents > div > a')
    # 수집한 데이터를 저장할 리스트 선언 
    # 링크 , 책이름 , 카테고리 =>category의 값에서 [국내도서]
    links, bookname, category = [],[],[]
    for lognum,e in enumerate(s):
        if 'https://' in e.attrs['href']:
            links.append(e.attrs['href'])
            contents = e.text.replace('예약판매','').replace('\n','').strip()
            catsearch1 = re.search('\[.+?\]',contents)
            cont1 = catsearch1.group()
            cont2 = re.sub('[\[\]]','',cont1)
            category.append(cont2)
            subject = contents.replace(cont1,'')
            bookname.append(subject)
            #global lognum
            #lognum += 1
    return links, bookname, category
```

```python
linksv, booknamev, categoryv = loopkyoboBooks('python',1)
linksv[:2] , booknamev[:2] ,  categoryv[:2]
```

```python
match = re.search(r'([\d,]+)', heading_text)
match
```

```python
total = int(match.group(1).replace(',', ''))
print(total)
```

```python
# 전체 페이지 구해보기
total_page = math.ceil(total/20)
total_page
```

```python
# 진행률 시각화 : tqdm 
# !pip install tqdm
# !pip list | grep tqdm
from tqdm import tqdm
```

```python
linksvv, booknamevv, categoryvv = [],[],[]
for page in tqdm(range(1,total_page+1),desc="Processing Page",unit="page"):
    linksv, booknamev, categoryv = loopkyoboBooks('python',page)
    linksvv += linksv
    booknamevv += booknamev
    categoryvv += categoryv
```
