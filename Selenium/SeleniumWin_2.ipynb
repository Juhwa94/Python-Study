```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
#Key Event처리하기 위한 모듈 
from selenium.webdriver.common.keys import Keys
from webdriver_manager.chrome import ChromeDriverManager
import time
from tqdm import tqdm
import os
from bs4 import BeautifulSoup
```

```python
#webdriver_manager를 사용해서 해당 버전을 자동으로 로딩 및 초기화
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
searchName = '카리나'
url=f'https://search.naver.com/search.naver?ssc=tab.image.all&where=image&query={searchName}'
driver.get(url)
time.sleep(2)
req = driver.page_source
```

```python
req[100:250]
```

```python
# <Event 제어>
# 현재 open된 크롬 브라우저에서 자바스크립트를 제어도 가능
# 현재 브라우저에 클릭 이벤트 적용
# 현재 브라우저 상태를 활성화 하기 위해서 클릭 
# By.CSS_SELECTOR -> CSS에서 제공하는 선택자를 의미
# body { }
body = driver.find_element(By.CSS_SELECTOR,'body')
body.click()
```

```python
body.send_keys(Keys.END)
time.sleep(2)
body.send_keys(Keys.HOME)
```

```python
# XPATH문법
# -> //TAG[@class=""]
# -> //TAG[@id=""]
#//*[@id="main_pack"]/section/div[1]/div/div/div[1]/div[489]/div/div/div/img

# find_element : 단일 요소 => find, select_one
# find_elements : 집합요소를 가져오기 위한 메서드 : find_all, select 
# 이미지들의 반복되는 class의 이름을 가져와서 하나씩 클릭하면서 이미지의 주소만
# 배열에 저장
imgs = driver.find_elements(By.XPATH,'//img[@class="_fe_image_tab_content_thumbnail_image"]')
imgs[1].get_attribute('src')
```

```python
# 이미지를 리스트에 저장해보자
imgUrl = []
for i , e in enumerate(imgs[:20]):
    time.sleep(2) #2초 잠시 멈춤
    print(f"i=>{i} : e => {e}")
    print('*'*30)
    imgUrl.append(e.get_attribute('src'))
```
