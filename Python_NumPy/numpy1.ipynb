파이썬에서 제공하는 time 함수를 사용해서 속도를 체크하는 함수를 각각 만든 후 동일한 size로 테스트 한다

```python
import time
size_v = 1000000
```

```python
x = np.array([1,1,1])
y = np.array([2,2,2])
print(x + y)
```

```python
x = [1,1,1]
y = [2,2,2]
print(x + y)
[ x[i] + y[i] for i in range(len(x)) ]
```

```python
def numpy_python():
    t1 = time.time() #start time
    x = np.arange(size_v)
    y = np.arange(size_v)
    z = x + y
    print(z)
    return time.time() - t1
```

```python
def list_python():
    t1 = time.time() #start time
    x = range(size_v)
    y = range(size_v)
    z = [ x[i] + y[i] for i in range(len(x)) ]
    print(z)
    return time.time() - t1
```

<h2 id="%EB%A6%AC%EC%8A%A4%ED%8A%B8%ED%98%95%EA%B3%BC-numpy-%EC%9D%98-%EB%B2%A1%ED%84%B0-%EC%97%B0%EC%82%B0-%EB%B9%84%EA%B5%90%ED%95%98%EA%B8%B0">리스트형과 numpy 의 벡터 연산 비교하기¶</h2><p></p>

```python
#리스트로 벡터 연산하기
# a + b = [a1 + b1 , a2 + b2 ,,,,, an + bn]
x = [1,2,3]
y = [4,5,6]
print(f'리스트와 리스트 덧셈은 ? {x+y}')
print(f'{x} + {y} = {[ x[i] + y[i] for i in range(len(x))]}')
print(f'{x} - {y} = {[ x[i] - y[i] for i in range(len(x))]}')
print(f'{x} * {y} = {[ x[i] * y[i] for i in range(len(x))]}')
```

```python
# numpy로 연산 해보기 
x  = np.array([1,2,3])
y  = np.array([4,5,6])
print('numpy 벡터연산')
print(f'{x} + {y} = {(x + y)}')
print(f'{x} - {y} = {(x - y).tolist()}')
print(f'{x} * {y} = {(x * y).tolist()}')
```

<p>3D텐서는 머신 러닝, 특히 시계열 데이터를 다루는 순환 신경망(Recurrent Neural Networks, RNN) 또는 LSTM(Long Short-Term Memory) 네트워크에서 널리 사용 된다.</p>

<p>X축 (Features) : 데이터의 가로 크기를 나타내며, 예를 들어, 주식 시장 데이터에서는 가격, 거래량, 개장 시간 등이 특성이 될 수 있다. 이미지일 경우 픽셀의 가로 길이가 된다.</p>

```python
def my_Aniface(face_color=[200, 200, 200], eye_color=[0, 0, 0]):
    height, width, num_channels = 64, 64, 3
    
    img = np.ones((height, width, num_channels), dtype=np.uint8) * 255
    
    img[20:50, 15:50] = face_color

    img[40:45, 22:27] = eye_color
    img[40:45, 38:43] = eye_color
    

    img[28:30, 30:35] = [255, 0, 0]
    

    img[50:60, 10:20] = [0, 0, 0] 
    img[50:60, 44:54] = [0, 0, 0] 
    
    return img


num_images = 3
height, width, num_channels = 64, 64, 3
tensor_4d = np.zeros((num_images, height, width, num_channels), dtype=np.uint8)


tensor_4d[0] = my_Aniface(face_color=[200, 200, 200], eye_color=[0, 0, 0])   
tensor_4d[1] = my_Aniface(face_color=[150, 200, 250], eye_color=[255, 0, 0]) 
tensor_4d[2] = my_Aniface(face_color=[100, 255, 100], eye_color=[0, 0, 255])  

print( tensor_4d.shape)


fig, axes = plt.subplots(1, num_images, figsize=(15, 5))

for i, ax in enumerate(axes):
    ax.imshow(tensor_4d[i], origin='lower')
    ax.grid(True)
    ax.axis('on')
    ax.set_title(f'MyAniFace {i+1}')

plt.tight_layout()
plt.show()
```

<h2 id="NumPy%EC%9D%98-reshape-%EA%B8%B0%EC%B4%88-%EC%9D%B4%ED%95%B4%ED%95%98%EA%B8%B0">NumPy의 reshape 기초 이해하기¶</h2>
reshape는 배열의 형태를 변경하는 NumPy의 핵심 기능이다.
전체 원소 수는 유지한 채 배열의 차원과 형태를 바꿀 수 있다.
reshape을 정의할 때 전체 원소 수는 변경되면 안된다.
-1을 사용할 때는 한 차원에서만 사용이 가능하다.

```python
# 1차원 배열 생성하기
arr = np.arange(1,7)
print(f"{arr.shape} : {arr}")
# 2 x 3 = [6]
arr_reshape = arr.reshape(2,3)
print(f"변환 배열 :\n {arr_reshape} \n shape:{arr_reshape.shape}")
```

```python
### -1 사용하여 자동 크기 계산
### 공식 :(총원소의 개수 / 고정된 차원) 
arr = np.arange(1,7)
# 2행은 고정, 열은 자동으로 계산
res = arr.reshape(2,-1)
print(res)
# 열은 3개로 고정, 행은 자동 계산
res = arr.reshape(-1,3)
print(res)
```

<p></p>

```python
arr = np.arange(10,100,10)
# 배열 arr의 총 원소 개수는 9개이며, 열을 3으로 고정했으므로 행은 9 / 3 = 3으로 계산된다.
# 결과적으로 3x3 행렬로 변환
result = arr.reshape(3,-1)
result
```

```python
arr = np.arange(1,9)
# 8 / 2 = 4 
# 4 x 2 행렬이됨 
result = arr.reshape(4, -1)
print(result)
```
